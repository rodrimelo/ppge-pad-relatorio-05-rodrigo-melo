## 🧰 Ferramentas permitidas — e o que não pode usar

**Permitido neste relatório:**

- `for` sobre **listas** e **tuplas**, inclusive **aninhado** (um `for` dentro de outro);
- `range(stop)`, `range(start, stop)` e `range(start, stop, step)`, inclusive com passo **negativo**;
- `enumerate(lista)` e `enumerate(lista, start=n)`, com desempacotamento `for i, item in ...`;
- listas e tuplas: indexação, indexação negativa, fatiamento, `len()`, `append()` e desempacotamento (`for a, b, c in lista_de_tuplas`);
- `break`, `continue` e a cláusula `else` do `for`;
- `if`, `elif`, `else`, `try`, `except`, `raise`;
- `print()`, `float()`, `int()`, `round()` e f-strings.

**Proibido neste relatório:**

- `while`;
- `zip()`;
- *list comprehensions* e expressões geradoras;
- dicionários e conjuntos;
- percorrer **strings** ou **dicionários** com `for` (aqui o `for` percorre apenas listas, tuplas, `range` e `enumerate`);
- funções próprias (`def` / `return`) — esse tópico será visto mais adiante;
- `sum()`, `max()`, `min()` e `sorted()` para substituir o laço (queremos ver o acumulador);
- NumPy, pandas ou qualquer biblioteca externa.

> 🚫 Se a solução usar uma estrutura proibida, a questão **não pontua**, mesmo que o resultado apareça. O objetivo é treinar o `for` com listas, tuplas, `range` e `enumerate`.

> 💡 **Regra de ouro:** evite `for i in range(len(lista))` quando `enumerate(lista)` resolve. Use `range` quando o problema é **contar** (meses, anos, parcelas) ou quando precisa de **índice vizinho** (`t - 1`).
